```python
import numpy as np

def clean_discipline(df):
    # 1. Drop the useless index column
    df = df.drop(columns=["Unnamed: 0", "discipline"], errors="ignore")

    # 2. Rename columns to be cleaner
    df = df.rename(columns={
        "Competitor":     "athlete_name",
        "DOB":            "dob",
        "Nat":            "country",
        "Results Score":  "results_score",
        "WIND":           "wind",
        "Pos":            "pos",
        "Venue":          "venue",
        "Date":           "date",
    })

    # 3. Parse dates properly
    df["date"]     = pd.to_datetime(df["date"], format="%d %b %Y", errors="coerce")
    df["dob"]      = pd.to_datetime(df["dob"],  format="%d %b %Y", errors="coerce")

    # 4. Calculate athlete age at time of performance
    df["age"] = ((df["date"] - df["dob"]).dt.days / 365.25).round(1)

    # 5. Filter to only legal wind readings for sprints
    # (keep all rows for field events where wind is NaN)
    df["wind_legal"] = df["wind"].isna() | (df["wind"] <= 2.0)

    # 6. Filter to 2021-2023 for our training window
    df_recent = df[df["year"].between(2021, 2023)].copy()

    # 7. Drop rows where Mark is missing
    df_recent = df_recent.dropna(subset=["Mark"])

    return df_recent

# Clean all disciplines
cleaned = {}
for key in dfs:
    cleaned[key] = clean_discipline(dfs[key])
    df_c = cleaned[key]
    print(f"{key}: {len(df_c)} rows, age missing: {df_c['age'].isna().sum()}, wind missing: {df_c['wind'].isna().sum()}")
```

```python
def clean_discipline(df):
    # 1. Drop the useless index column
    df = df.drop(columns=["Unnamed: 0", "discipline"], errors="ignore")

    # 2. Rename columns to be cleaner
    rename_map = {
        "Competitor":    "athlete_name",
        "DOB":           "dob",
        "Nat":           "country",
        "Results Score": "results_score",
        "Pos":           "pos",
        "Venue":         "venue",
        "Date":          "date",
    }
    # only rename WIND if it exists
    if "WIND" in df.columns:
        rename_map["WIND"] = "wind"

    df = df.rename(columns=rename_map)

    # 3. If no wind column exists, create it as NaN
    if "wind" not in df.columns:
        df["wind"] = np.nan

    # 4. Parse dates properly
    df["date"] = pd.to_datetime(df["date"], format="%d %b %Y", errors="coerce")
    df["dob"]  = pd.to_datetime(df["dob"],  format="%d %b %Y", errors="coerce")

    # 5. Calculate athlete age at time of performance
    df["age"] = ((df["date"] - df["dob"]).dt.days / 365.25).round(1)

    # 6. Flag legal wind readings (NaN = field event = always legal)
    df["wind_legal"] = df["wind"].isna() | (df["wind"] <= 2.0)

    # 7. Filter to 2021-2023 training window
    df_recent = df[df["year"].between(2021, 2023)].copy()

    # 8. Drop rows where Mark is missing
    df_recent = df_recent.dropna(subset=["Mark"])

    return df_recent

# Clean all disciplines
cleaned = {}
for key in dfs:
    cleaned[key] = clean_discipline(dfs[key])
    df_c = cleaned[key]
    print(f"{key}: {len(df_c)} rows, age missing: {df_c['age'].isna().sum()}, wind missing: {df_c['wind'].isna().sum()}")
```

```python
import os

PROCESSED_DIR = r"C:\Users\rayen\athletics-predictor\data\processed"
os.makedirs(PROCESSED_DIR, exist_ok=True)

# Save cleaned files
for key, df in cleaned.items():
    out_path = os.path.join(PROCESSED_DIR, f"{key}_clean.csv")
    df.to_csv(out_path, index=False)
    print(f"Saved {key}_clean.csv")

print("\nAll cleaned files saved to data/processed/")
```
